```python
# Detect potential outliers using the IQR (Interquartile Range) method
# IQR helps identify values that are unusually far from the normal range

for column in outlier_columns:

    # Calculate the 25th percentile
    Q1 = df[column].quantile(0.25)

    # Calculate the 75th percentile
    Q3 = df[column].quantile(0.75)

    # Calculate the Interquartile Range
    IQR = Q3 - Q1

    # Calculate lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Count values outside the normal IQR range
    outlier_count = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(f"{column}: {outlier_count} potential outliers")
```

```python
# Create a function to convert European-style numbers into normal numeric values
# Example:
# "28,57"      -> 28.57
# "$2.084,25"  -> 2084.25
# "70,68%"     -> 70.68

def clean_number(value):

    # Keep missing values as missing
    if pd.isna(value):
        return np.nan

    # Convert value into text
    value = str(value)

    # Remove dollar symbol
    value = value.replace("$", "")

    # Remove percentage symbol
    value = value.replace("%", "")

    # Remove thousand separator
    value = value.replace(".", "")

    # Replace decimal comma with decimal point
    value = value.replace(",", ".")

    # Convert cleaned text into a numeric value
    return pd.to_numeric(value, errors="coerce")
```

```python
# List of columns that use comma decimals, currency symbols, or percentage symbols

formatted_numeric_columns = [
    "Avg_Daily_Sales",
    "Days_of_Inventory",
    "Unit_Cost_USD",
    "Last_Purchase_Price_USD",
    "Total_Inventory_Value_USD",
    "SKU_Churn_Rate",
    "Order_Frequency_per_month",
    "Supplier_OnTime_Pct",
    "Audit_Variance_Pct",
    "Demand_Forecast_Accuracy_Pct"
]

# Restore these columns from the raw dataset
# Then convert them correctly into numeric format

for column in formatted_numeric_columns:
    df[column] = raw_df[column].apply(clean_number)

print("Formatted numeric columns cleaned successfully.")
```

```python
# Calculate Available-to-Promise inventory
# ATP represents the stock that is actually available for customer demand

df["ATP"] = (
    df["Quantity_On_Hand"]
    - df["Quantity_Reserved"]
    - df["Quantity_Committed"]
)

# Display sample values
df[
    [
        "SKU_ID",
        "Quantity_On_Hand",
        "Quantity_Reserved",
        "Quantity_Committed",
        "ATP"
    ]
].head()
```

```python
# Create an Oversold Flag
# An SKU is oversold when reserved + committed stock exceeds physical stock

df["Oversold_Flag"] = np.where(
    df["ATP"] < 0,
    1,
    0
)

# Count oversold SKUs
print("Oversold SKUs:", df["Oversold_Flag"].sum())
```

```python
# Create a Below Safety Stock Flag
# We compare ATP with Safety Stock because ATP represents actual sellable stock

df["Below_Safety_Stock_Flag"] = np.where(
    df["ATP"] < df["Safety_Stock"],
    1,
    0
)

# Count SKUs below safety stock
print(
    "SKUs Below Safety Stock:",
    df["Below_Safety_Stock_Flag"].sum()
)
```

```python
# Create a Reorder Flag
# If ATP is at or below the reorder point,
# the SKU should be reviewed for replenishment

df["Reorder_Flag"] = np.where(
    df["ATP"] <= df["Reorder_Point"],
    1,
    0
)

# Count SKUs requiring reorder review
print(
    "SKUs Requiring Reorder Review:",
    df["Reorder_Flag"].sum()
)
```

```python
# Create sellable ATP for coverage calculation
# Negative ATP cannot provide customer coverage,
# so negative values are treated as zero for this metric

df["Sellable_ATP"] = df["ATP"].clip(lower=0)

# Calculate Sellable Coverage Days
# This tells us approximately how many days current sellable stock can support demand

df["Sellable_Coverage_Days"] = np.where(
    df["Avg_Daily_Sales"] > 0,
    df["Sellable_ATP"] / df["Avg_Daily_Sales"],
    np.nan
)

# Round the result for easier interpretation
df["Sellable_Coverage_Days"] = df["Sellable_Coverage_Days"].round(2)

df[
    [
        "SKU_ID",
        "ATP",
        "Avg_Daily_Sales",
        "Sellable_Coverage_Days"
    ]
].head()
```

```python
# Create an Expired Flag
# This identifies stock whose expiry date is before the analysis date

df["Expired_Flag"] = np.where(
    df["Expiry_Date"].notna()
    & (df["Expiry_Date"] < analysis_date),
    1,
    0
)

print("Expired SKUs:", df["Expired_Flag"].sum())
```

```python
# Create a flag for stock expiring within the next 30 days
# Already expired stock is excluded from this flag

df["Expiry_30_Days_Flag"] = np.where(
    df["Expiry_Date"].notna()
    & (df["Expiry_Date"] >= analysis_date)
    & (df["Expiry_Date"] <= analysis_date + pd.Timedelta(days=30)),
    1,
    0
)

print(
    "SKUs Expiring Within 30 Days:",
    df["Expiry_30_Days_Flag"].sum()
)
```

```python
# Create a Potential Dead Stock Flag
# Stock held for more than 180 days is treated as potential dead stock

df["Dead_Stock_Flag"] = np.where(
    df["Stock_Age_Days"] > 180,
    1,
    0
)

print(
    "Potential Dead Stock SKUs:",
    df["Dead_Stock_Flag"].sum()
)
```

```python
# Calculate inventory value that may expire within the next 30 days
# This represents immediate financial risk from near-expiry inventory

expiry_30_value = df.loc[
    df["Expiry_30_Days_Flag"] == 1,
    "Total_Inventory_Value_USD"
].sum()


# Calculate value of already expired inventory
expired_value = df.loc[
    df["Expired_Flag"] == 1,
    "Total_Inventory_Value_USD"
].sum()


# Calculate potential dead stock value
dead_stock_value = df.loc[
    df["Dead_Stock_Flag"] == 1,
    "Total_Inventory_Value_USD"
].sum()


print("===== FINANCIAL RISK SUMMARY =====")

print(
    "Value Expiring Within 30 Days (USD):",
    round(expiry_30_value, 2)
)

print(
    "Expired Inventory Value (USD):",
    round(expired_value, 2)
)

print(
    "Potential Dead Stock Value (USD):",
    round(dead_stock_value, 2)
)
```

```python
# Create warehouse-level risk summary
# This helps identify which warehouse needs the most management attention

warehouse_insights = df.groupby(
    ["Warehouse_ID", "Warehouse_Location"]
).agg(

    Total_SKUs=("SKU_ID", "count"),

    Inventory_Value=(
        "Total_Inventory_Value_USD",
        "sum"
    ),

    Below_Safety_SKUs=(
        "Below_Safety_Stock_Flag",
        "sum"
    ),

    Reorder_SKUs=(
        "Reorder_Flag",
        "sum"
    ),

    Oversold_SKUs=(
        "Oversold_Flag",
        "sum"
    ),

    Expiring_30_Days=(
        "Expiry_30_Days_Flag",
        "sum"
    )

).reset_index()


# Sort warehouses by safety-stock risk
warehouse_insights = warehouse_insights.sort_values(
    by="Below_Safety_SKUs",
    ascending=False
)

warehouse_insights
```

```python
# Identify the warehouse with the highest number of SKUs below safety stock

highest_risk_warehouse = warehouse_insights.iloc[0]

print(
    "Warehouse with Highest Safety Stock Risk:",
    highest_risk_warehouse["Warehouse_ID"],
    "-",
    highest_risk_warehouse["Warehouse_Location"]
)

print(
    "SKUs Below Safety Stock:",
    highest_risk_warehouse["Below_Safety_SKUs"]
)
```

```python
# Create supplier performance summary
# Lower on-time delivery percentages indicate weaker supplier reliability

supplier_insights = df.groupby(
    "Supplier_Name"
).agg(

    Total_SKUs=("SKU_ID", "count"),

    Avg_OnTime_Pct=(
        "Supplier_OnTime_Pct",
        "mean"
    ),

    Avg_Lead_Time_Days=(
        "Lead_Time_Days",
        "mean"
    ),

    Inventory_Value=(
        "Total_Inventory_Value_USD",
        "sum"
    )

).reset_index()


# Round numerical values for easier interpretation
supplier_insights["Avg_OnTime_Pct"] = (
    supplier_insights["Avg_OnTime_Pct"].round(2)
)

supplier_insights["Avg_Lead_Time_Days"] = (
    supplier_insights["Avg_Lead_Time_Days"].round(2)
)


# Sort from weakest to strongest on-time delivery
supplier_insights = supplier_insights.sort_values(
    by="Avg_OnTime_Pct",
    ascending=True
)

supplier_insights
```

```python
# Identify the supplier with the lowest average on-time delivery rate

weakest_supplier = supplier_insights.iloc[0]

print(
    "Supplier with Lowest On-Time Delivery:",
    weakest_supplier["Supplier_Name"]
)

print(
    "Average On-Time Delivery %:",
    weakest_supplier["Avg_OnTime_Pct"]
)

print(
    "Average Lead Time:",
    weakest_supplier["Avg_Lead_Time_Days"],
    "days"
)
```
